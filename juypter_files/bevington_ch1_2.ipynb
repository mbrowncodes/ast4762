```python
# illegitimate errors:
  # originate from mistsake/blunder in measure or computing
  # can be corrected by carefully repeating operations

# accuracy vs precision:
  # a: how close result is to true value
  # p: reproducibility of result
  # uncertainity/error --> refer to precision of result found
  # absolute precision= mag of uncertainity in result in same unit of result
  # relative precision= uncertainty in terms of fraction of value of result

# systematic errors
  # accuracy depend on how well we control systematic errors 
  # these are things that make the results different from true values
  # not easy to see, not easily studied (faulty calibration of equip)
  # plan experiment : understand & reduce source of systematic errors


# random errors
  # precision depend on how well we overcome random errors
  # flux in observation giving different results each time
  # accuracy imply = precision (depend on random error)
  # if from instrument --> reduce by using more precisie measuring instruments 
  # if from statistical fluctuations--> take more measurements
```

```python
# Significant Figures and Roundoff
  # precision of result --> implied by number sig fig
  # leftmost digit: most signifiicant digit
  # no decimal pt: rightmost nonzero digit least significant
  # decimal pt: rightmost digit least significant (even if 0)
  # all digit between least and most sig dig are counted as sig dig
  # write in scientific notation if all digits are sig (ie 1010 looks like 3 sig fig)
  # experimental result: # sig fig = 1 extra than dictated by experimental precision
  # --> reason: avoid errors caused by rounding errors (in calculations)
  # if it equals 1/2 aka 0.5 increment least sig dig only iff odd
  # only round final values
  
```
