```python
def price_good (p, d, c):
    """
    Input: 
    p: price
    d: discount (percentage, 40% is 0.4)
    c: amount of copies
    Output:
    price after discount
    shipping cost for # of copies
    total cost
    """

    p = float(p) 
    d = int(d)
    if d <= 0:
        print ("That discount is not valid")
        total_discounted = p
    elif d > 0:
        d = d*0.01 #change discount into decimal percentage
        total_discount = p - (p * d) 
    new_price = total_discounted
    print(new_price)
        #print("the total after the discount applied is", total_discounted)
    c = int(c) #can only have 2 copies not 2.2 , make int

    for k in range(c+1):
        if k == 1 :
            shipping = 3 #for only one copy cost is $3
        elif k > 1:
            shipping = 3 + (c-1)*0.75 #number of copies is $3+0.75c per copy
            
    total_spent = total_discounted + shipping 
    return (total_spent)


    

    
            
```
