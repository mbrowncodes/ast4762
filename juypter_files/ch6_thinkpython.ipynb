```python
#recursive
def factorial(n):
    if n ==0:
        return 1 # value of 0 factorial
    else:
        recurse= factorial(n-1)
        result = n*recurse
        return result

        #ie, n=3 , n(n-1)(n-2)
        # 3 !=0 do recurse, then 2 != 0 do recurse, then 1 != 0 do recurse
factorial(3)
```
