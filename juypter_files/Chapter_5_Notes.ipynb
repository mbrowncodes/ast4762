```python
# recursion
# legal to have function call its-self


def countdown(n):
    if n<= 0:
        print("blastoff")
    else:
        print(n) #prints the value of n that is pos or nonzero
        countdown(n-1) # prints each n-1 value until it equlas 0, then the if statement runs
countdown(3)
# ex of recursive, 


def print_n(s, n):
    '''
    printss a string a number of times
    input: s = string
    n = number
    output: string(s) prints (n) times
    '''
    if n<= 0: #doesnt print because 0 or negative isnt valid
        return #EXITS THE FUNCTION
    print(s) # if not 0 or neg, print the string
    print_n(s, n-1) #keep printing the string until n=0 so that the function returns/ends
print_n('lol', 3)
```
