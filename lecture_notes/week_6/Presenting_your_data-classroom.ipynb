### 2D plots:

```python
#let's make a data set 
x = np.arange( 0, 100, 0.01 )
y = np.sin( 2 * np.pi * x )
```

```python
#we can make a fast, dirty plot:
dm = plt.plot( x, y )  
```

### While this is ‘a’ plot, it is not ideal. What are x and y showing? Also, it is a bit small.  Let’s add titles (xlabel/ylabel) and increase the size of the figure (figsize). While at it, let’s also change the color of the line to red (color). You can find all colors [here](https://matplotlib.org/3.1.0/gallery/color/named_colors.html)


```python
#define how big the plot will appear:
plt.figure( figsize = ( 12,4 ) )

#plot it using a red, solid line
plt.plot  ( x, y, color = 'firebrick' ) # (or 'r')

# add a label on the x-axis
plt.xlabel( 't' )

#add a label on the y-axis
dm = plt.ylabel( 'Sin[ 2 $\pi$ t ] ' )
```

### Let’s change the linestyle to a dashed line. You can find all linestyles [here](https://matplotlib.org/3.1.0/gallery/lines_bars_and_markers/linestyles.html) . We will also increase the fontsize of our labels to be able to actually read them from the back of the room, and zoom in in 2 periods.

```python
plt.figure( figsize = (5, 5 ) )
plt.plot  ( x, y, color = 'red', linestyle = '--' ) 
plt.xlabel( 't'                 , fontsize = 20 )
plt.ylabel( 'Sin[ 2 $\pi$ t ] ' , fontsize = 20 )

#limit x axis
plt.xlim( 0, 2.1 )

#make tick labels legible:
plt.yticks( fontsize = 16 )

#note that if I skip the np.arange all labels will be printed and they will 
#make a crowded/overlapping x axis mess
plt.xticks( np.arange(0.0,2.1,0.5), fontsize = 16 )


#add a title on the plot and make sure it has big enough font

dm = plt.title('My sin function', fontsize = 20)
```

### One more thing that I like to change to make the plot better legible is the orientation and size of [tickmarks](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.tick_params.html) themselves. 

```python

plt.figure( figsize = (5, 5 ) )
plt.plot  ( x, y, color = 'red', linestyle = '--' ) 
plt.xlabel( 't'                 , fontsize = 16 )
plt.ylabel( 'Sin[ 2 $\pi$ t ] ' , fontsize = 16 )

#limit x axis
plt.xlim( 0, 2.1 )

#make tick labels legible:
plt.yticks( fontsize = 16 )

#note that if I skip the np.arange all labels will be printed and they will 
#make a crowded/overlapping x axis mess
plt.xticks( np.arange(0.0,2.1,0.5), fontsize = 16 )


#add a title on the plot and make sure it has big enough font

plt.title('My sin function', fontsize = 16)


#access and change the tick parameters for this plot:
dm = plt.tick_params(axis='both',which='major',length=8, width=1.5,direction='in')
#plt.tick_params(axis='both',which='minr',length=8, width=3,direction='in')
```

### Let's move to another dataset now

```python
x2  = np.arange( 20 )
y2a = 2.1*x2 + 3.1
y2b = 3.0*x2 + 2.5
```

#### plot both data in the same plot using different colors and linestyles. Add a [legend](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.legend.html) to indicate what is what. Note that we need to  define the labels first otherwise there is nothing for legend to display

```python
plt.figure( figsize = (5, 5 ) )
plt.plot  ( x2, y2a, color = 'firebrick', linestyle = '--' , 
           linewidth = 2, label = 'model 1')
plt.plot  ( x2, y2b, color = 'cornflowerblue', linestyle = '-.' , 
           linewidth = 2, label = 'model 2')

#fix tick label size
plt.yticks( np.arange(10, 61, 20), fontsize = 16 )
plt.xticks( np.arange(0, 21, 5) , fontsize = 16 )


#add a title on the plot and make sure it has big enough font

plt.title('Two lines', fontsize = 16)

#access and change the tick parameters for this plot:
plt.tick_params(axis='both',which='major',length=8, width=1.5,direction='in')

#add a legend
dm = plt.legend( fontsize = 12, loc = 'upper left')
```

### If you want to add more flexibility to your plots, and start making more advanced/ multiple plots you can do more object-oriented plots using the 'ax' 

```python
#let's make a basic plot of x vs y for this random sinusoidal data:
x = np.linspace( 0, 15, 1000 )
y = np.sin( x )
```

```python
fig, ax = plt.subplots( figsize = (4, 3) )  # create a figure object

ax.plot( x, y , linewidth = 4)                     #make the plot 
ax.set_ylabel( 'Y', color='firebrick', fontsize=20 )    # add X and Y labels and make them red
ax.set_xlabel( 'X', color='blue', fontsize=20 )
ax.tick_params( axis="both", labelsize=18 )       # change the size of your tick labels
```

### this method offers you more flexibility than the plt--- for example, you can use double y axis in a plot:  (note: same principles for double x axis)

```python
#make a new x axis:
x = np.linspace(0., 10., 100)

fig = plt.figure( figsize = (6,5) )

#I will start with ax1 
ax1 = fig.add_subplot(1, 1, 1)

ax1.plot( x, x ** 2, color= 'blue' , linewidth = 3 )
#I can rotate the y label 
ax1.set_ylabel( 'x$^2$', color='blue', fontsize=24, rotation = 0 )
ax1.set_xlabel('x',fontsize=20)
ax1.tick_params( axis="both", labelsize=18, length = 7 , direction = 'in' )

#but then I want a secondary y axis, so I need to make an ax1 twin named ax2
ax2 = ax1.twinx()

ax2.plot( x, 1/x, color='red')
#and this one I will leave the label as is
ax2.set_ylabel('$ \\frac{1}{x} $', color='red',fontsize=24) #, rotation = 0)
dm = ax2.tick_params( axis="y", labelsize=18 , length = 7 , direction = 'in' )


# want two things on same graph that share x axis
# twin one of the axis
# define new axis by twinning axis
```

```python
#notice how different that is from the 'normal' 2 in 1 plot:
plt.plot( x, x ** 2, color= 'blue' , linewidth = 3)
dm = plt.plot(x, 1/x, color='red')
```

### with the ax method you can start making more complex plots

```python
# I start by defining that I want 2 line 1 columns plot
fig, ( ax1, ax2 ) = plt.subplots(2, figsize = (10, 10 )) # this is 2,1 figures

fig.suptitle('Vertically stacked scatter plots example' , fontsize = 24 )

#start with the first one:

ax1.plot( x , x**2 )
ax1.set_xlabel( ' x ' , fontsize = 14 )
ax1.set_ylabel( ' $x^2$' , fontsize = 14 )

#you can define what your tickmarks will be via the set_xticks (or y) and _xticklabels
ax1.tick_params( axis="both", labelsize=18, length = 7 , direction = 'in' )

# go to the second one:

ax2.plot( x, 1/x , color = 'red' ) 
ax2.set_xlabel( ' x ' , fontsize = 14 )
ax2.set_ylabel( r' $\frac{1}{x}$ ' , fontsize = 14 )

dm = ax2.tick_params( axis ='both', which ='both', length = 7 , direction = 'in' , labelsize=14 )
```

```python
# but I realized I needed 2 columns 1 line plot to make it look better
fig, ( ax1, ax2 ) = plt.subplots(1, 2, figsize = (12, 6 )) # this is 1,2 figure

fig.suptitle('Horizontally stacked scatter plots example' , fontsize = 24 )

#start with the first one:

ax1.plot( x , x**2 )
ax1.set_xlabel( ' x ' , fontsize = 14 )
ax1.set_ylabel( ' $x^2$' , fontsize = 14 )

#you can define what your tickmarks will be via the set_xticks (or y) and _xticklabels
ax1.tick_params( axis="both", labelsize=18, length = 7 , direction = 'in' )

# go to the second one:

ax2.plot( x, 1/x , color = 'red' ) 
ax2.set_xlabel( ' x ' , fontsize = 14 )
ax2.set_ylabel( r' $\frac{1}{x}$ ' , fontsize = 14 )

dm = ax2.tick_params( axis ='both', which ='both', length = 7 , direction = 'in' , labelsize=14 )
```

### often you will need to use contours to highlight a 3rd dimension in your data; or colorbars to show the levels that your data/image shows (we have already seen a number of colorbars with imshow() )

```python
fig, ax = plt.subplots(1, 1, figsize=(7, 7))
#make dataset
delta = 0.025
x    = np.arange( -5, 5, 0.1 )
y    = np.arange( -5.0, 5.0, delta )
xx, yy = np.meshgrid( x, y , sparse=True )

z = np.sin( xx**2 + yy**2 ) / ( xx**2 + yy**2 )

#normalize the data for colorbar and let code know you use 256 colors
bounds = np.linspace( np.min( z ) , np.max( z ) + .1, 100)
norm = colorsplt.BoundaryNorm(boundaries=bounds, ncolors=256)

#define the filled contour plot
pcm = ax.contourf(x, y ,z ,100, cmap='RdYlBu')

#define that you want its colorbar
cbar = fig.colorbar( pcm,fraction=0.045, pad=0.01 )

#add the cosmetic stuff:
cbar.set_label('random contour',fontsize = 20 )
cbar.ax.tick_params(labelsize = 20)

ax.tick_params( axis ='both', which ='both', length = 7 , direction = 'in', width = 3 )
ax.set_xticklabels( -6. + np.arange( 6 ) *2, fontsize = 20)
dm = ax.set_yticklabels( -6. + np.arange( 6 ) *2, fontsize = 20)
```

```python
# or you can do a contour plot that is not filled in:

fig, ax = plt.subplots( figsize = ( 12, 8 ) )

CS = ax.contour( x, y, z )
ax.clabel( CS, inline=1, fontsize=12 )
ax.set_title( 'Simple contour plot with labels' , fontsize = 20 )

ax.set_xlabel( 'X' , fontsize = 24 )
ax.set_ylabel( 'Y' , fontsize = 24 )
ax.tick_params( axis="both", labelsize=20 ) 
ax.tick_params( axis ='both', which ='both', length = 7 , direction = 'in' )
```

```python
# you can use color maps:
fig, ax = plt.subplots( figsize = ( 12, 8 ) )

CS = ax.contour( x, y, z, cmap = 'rainbow',
                )

ax.clabel( CS, fontsize = 14, inline = True )
ax.set_title( 'Crazy lines', fontsize = 24)

ax.set_xlabel( 'X' , fontsize = 24 )
ax.set_ylabel( 'Y' , fontsize = 24 )
ax.tick_params( axis="both", labelsize=20 ) 
ax.tick_params( axis ='both', which ='both', length = 7 , direction = 'in' )
```

```python
# you can also manually define the properties of your lines:
delta = 0.025
x = np.arange( -3.0, 3.0, delta )
y = np.arange( -2.0, 2.0, delta )
X, Y = np.meshgrid(x, y)

Z1 = np.exp(-X**2 - Y**2)
Z2 = np.exp(-(X - 1)**2 - (Y - 1)**2)
Z  = (Z1 - Z2) * 2

fig, ax = plt.subplots( figsize = ( 12, 8 ) )

CS = ax.contour( X, Y, Z, 6,
                linewidths = np.arange(1, 7, 1),
                colors = ('red', 'green', 'blue', 'orange', 'magenta', 'gray'),
                )

ax.clabel( CS, fontsize = 14, inline = True )
ax.set_title( 'Crazy lines', fontsize = 24)

ax.set_xlabel( 'X' , fontsize = 24 )
ax.set_ylabel( 'Y' , fontsize = 24 )
ax.tick_params( axis="both", labelsize=20 ) 
ax.tick_params( axis ='both', which ='both', length = 7 , direction = 'in' )
```

### How do you ensure that colorblind people can see your plots as they should be seen? 
#### community has created number of resources that you can use to plot colorblind-friendly plots; here I show just one from an old colleague

<img src="colorblind_friendly.png" width=550 height=550 />

source: https://personal.sron.nl/~pault/

```python
from tol_colors import tol_cset
```

```python
schemes = tol_cset()

fig, axes = plt.subplots(ncols=len(schemes))
fig.subplots_adjust(top=0.92, bottom=0.02, left=0.02, right=0.92)
for ax, scheme in zip(axes, schemes):
        cset = tol_cset(scheme)
        names = cset._fields
        colors = list(cset)
        for name, color in zip(names, colors):
            ax.scatter([], [], c=color, s=80, label=name)
        ax.set_axis_off()
        ax.legend(loc=2)
        ax.set_title(scheme)
plt.show()
```

```python
x = np.arange( 10 )
y = x ** 2.
z = x ** 0.5
w = x ** 1.5


clrs = tol_cset( colorset= 'bright' )

#add your plot here:
plt.figure( figsize =( 12, 7 ) )

plt.plot( x, y, linewidth =3 , color = clrs[0] )
plt.plot( x, z, linewidth =3 , color = clrs[3] )
plt.plot( x, w, linewidth =3 , color = clrs[5] )
```
