```python
import numpy as np

t = np.arange(0., 5., 0.2) # sample of time in 200ms intervals, time in s

plt.plot(t, t, 'r--', t, t**2, 'bs', t, t**3, 'g^')
# 3 functions with different colors and shapes of value dots
# x value of t then y value of t
plt.show()
```

```python
# sometimes have format where data lets you access variables with strings
# ie structured arrays or pandas.DataFrame

# give object data keyword argument

data = {'a': np.arange(50), 
        'c': np.random.randint(0,50,50),
        'd': np.random.randn(50)}
#np.random makes random numbers and samples data
# randint give integer values
# randn float point numbers 
# a 1d array, c 2d array, d 1d array

data['b'] = data['a'] + 10 * np.random.randn(50)
data['d'] = np.abs(data['d'] * 100)

plt.scatter('a', 'b', c='c', s='d', data = data)
plt.xlabel('entry a')
plt.ylabel('entry b')
plt.show()
```

```python
# plot with categorical variables 
names = ['group_a', 'group_b', 'group_c']
values = [1, 10, 100] 

plt.figure(figsize=(9, 3))

plt.subplot(131) #1 row, 3 columns, 1 = index of current subplot
plt.bar(names, values) #bar graph #makes subplot from data from variable names & values
plt.subplot(132)
plt.scatter(names,values) #scatter plot
plt.subplot(133)
plt.plot(names, values) # regular polynomial 
plt.suptitle('Categorical Plotting') # title of all subplots
plt.show()
```
