```python
print("Problem 1")

def find_sample (mean, sigma, N):
    """
    with a given standard deviation, mean, and N count value
    an array that is a sample of given values
    """
    samples = np.random.normal(loc = sigma, scale = sigma, size = N)
    # find normal gaussian, with random values
    return samples 
    
sigma = 13 #given value
mean = 55 #given value 
output = []
```

```python
print("Problem 2")

N = 10 #number of counts


for i in np.arange(10):
    # 10 being the number we want to loop through
    sample_values = find_sample(55, 13, 10) #go through function with given values
    mean_values = np.mean(sample_values) #find mean of samples
    sigma_values = np.std(sample_values) #find standard deviation
    output_values = [i, sigma_values, mean_values]
    output.append(output_values) # append these values into one array


    print(np.size(output_values))
   
#np.columnstack
```
