```python
import linfit as L
import numpy as np
import matplotlib.pyplot as plt
import astropy.io.fits as fits
```

```python
print("Problem 1")

#equations values
b = 1.2 #y intercept
m = 3.2 #slope



# read in data from file
model_1 = np.loadtxt("practicum3_1.dat", skiprows = 2,   max_rows = 100)
# picks out data for model 1 (no header, until the end of values)
model_2 = np.loadtxt("practicum3_1.dat", skiprows = 105)
# picks out data for model 2 (none of model 1 and no headers)

#model 1 data 
x1 = model_1[:, 0] # all values in first column (x column)
y1 = model_1[:, 1] # all values in 2nd column (y column)
#model 2 data
x2 = model_2[:, 0]
y2 = model_2[:, 1]


#plot model 1 
plt.figure(figsize=(7,7))
plt.plot(x1, y1, ".", label="f(x) = 3.2x + 1.2")
plt.title("Model 1")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()

```

```python
rel = L.linfit(y1, x1, 0.5) # linear fit to model 1 data, with y uncertainty 
print(rel[0:1])
print(f"Y-Intercept: {rel[0]:.4f} +- {rel[2]:.4f}", "Real Value:", b)
print(f"Slope:       {rel[1]:.4f} +- {rel[3]:.4f}", "Real Value:", m)

# plot from before)
plt.figure(figsize=(7,7)) 
plt.plot(x1, y1, ".", label="Data")
plt.plot(x1, rel[1] * x1 + rel[0], label ="Linear fit")
plt.title("Model 1")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()
```
