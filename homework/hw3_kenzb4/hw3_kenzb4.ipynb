```python



def squareplot(low, high, pt_num, saveplot=False):
    x = np.linspace(low, high, pt_num) #makes array from input graph values, includes
    # highest value of x in it 
    y = square(x) #makes new array that is square of x array values

    plt.figure(  figsize=(8,5)  ) #make figure readable size
    plt.plot(x, y)  # plot x vs y 
    plt.title("Square vs Original Data") 
    plt.xlabel("X")
    plt.ylabel("Y")

# example squareplot(0, 10, 5)

squareplot(1, 7, 5)
```
